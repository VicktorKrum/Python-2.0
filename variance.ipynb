```python
data = [25, 37, 15, 36, 92, 28, 33, 40]
```

```python
sorted_data = sorted(data)
```

```python
n = len(data)
```

```python
mean = int(sum(data)/n)
```

```python
'''summation=0
for i in range (0,n):
  summation += ((data[i]-mean)**2)
sd = (1/n)*(summation**(1/2))'''
```

```python
summation=0
for i in range (0,n):
  summation += ((data[i]-mean)**2)
var = summation/n
```

variance is square of standard deviation

```python
sd = var**(1/2)
```

```python
print('Std dev =',sd)
print('var = ', var)
```
